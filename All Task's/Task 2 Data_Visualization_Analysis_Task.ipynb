# Data Visualization & Analysis
**Internship Task Solution**

This notebook demonstrates exploratory data analysis and visualization using **Matplotlib** and **Seaborn**. It includes:
- Histograms with Kernel Density Estimation (KDE)
- Scatterplots with Trendlines and Categorical Groupings
- Annotated Correlation Matrix Heatmaps
- Advanced Box plots with Data Overlays

### 1. Import Libraries & Configure Global Settings

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set global theme and aesthetic parameters
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['font.sans-serif'] = "DejaVu Sans"
plt.rcParams['figure.dpi'] = 150
```

### 2. Generate / Load Dataset
*Note: To use a custom CSV, replace the synthetic dataset with `df = pd.read_csv('your_file.csv')`.*

```python
# Seed for reproducibility
np.random.seed(42)

data = {
    'Age': np.random.randint(22, 60, size=200),
    'Salary': np.random.randint(35000, 120000, size=200),
    'Spending_Score': np.random.randint(1, 100, size=200),
    'Years_Experience': np.random.randint(1, 35, size=200),
    'Department': np.random.choice(['IT', 'HR', 'Sales', 'Marketing'], size=200)
}
df = pd.DataFrame(data)

# Add continuous trend relationship for realistic plotting
df['Salary'] = df['Salary'] + (df['Years_Experience'] * 1500)

print("--- First 5 Rows ---")
display(df.head())

print("\n--- Dataset Summary Statistics ---")
display(df.describe())
```

### 3. Data Visualization Dashboard
Creating a multi-panel dashboard containing Histograms, Scatterplots, Correlation Matrices, and Boxplots.

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle('Data Visualization & Analysis Dashboard', fontsize=18, fontweight='bold', y=0.98)

# ------------------------------------------
# CHART 1: HISTOGRAM WITH KDE
# ------------------------------------------
sns.histplot(
    data=df, 
    x='Salary', 
    kde=True, 
    color='#2b5c8f', 
    bins=20, 
    ax=axes[0, 0]
)
axes[0, 0].set_title('1. Salary Distribution (Histogram + KDE)', fontsize=13, fontweight='bold')
axes[0, 0].set_xlabel('Salary ($)')
axes[0, 0].set_ylabel('Frequency')

# ------------------------------------------
# CHART 2: SCATTERPLOT WITH REGRESSION LINE
# ------------------------------------------
sns.scatterplot(
    data=df, 
    x='Years_Experience', 
    y='Salary', 
    hue='Department', 
    style='Department',
    s=80, 
    alpha=0.85, 
    ax=axes[0, 1]
)
sns.regplot(
    data=df, 
    x='Years_Experience', 
    y='Salary', 
    scatter=False, 
    ax=axes[0, 1], 
    color='black', 
    line_kws={'linestyle': '--', 'linewidth': 1.5}
)
axes[0, 1].set_title('2. Experience vs. Salary (Scatter Plot)', fontsize=13, fontweight='bold')
axes[0, 1].set_xlabel('Years of Experience')
axes[0, 1].set_ylabel('Salary ($)')

# ------------------------------------------
# CHART 3: ANNOTATED CORRELATION MATRIX
# ------------------------------------------
numeric_cols = df.select_dtypes(include=[np.number])
corr = numeric_cols.corr()

sns.heatmap(
    corr, 
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', 
    linewidths=0.8, 
    cbar=True, 
    square=True, 
    ax=axes[1, 0]
)
axes[1, 0].set_title('3. Correlation Matrix Heatmap', fontsize=13, fontweight='bold')

# ------------------------------------------
# CHART 4: BOXPLOT WITH STRIPPLOT OVERLAY
# ------------------------------------------
sns.boxplot(
    data=df, 
    x='Department', 
    y='Spending_Score', 
    palette='Set2', 
    ax=axes[1, 1]
)
sns.stripplot(
    data=df, 
    x='Department', 
    y='Spending_Score', 
    color='black', 
    alpha=0.3, 
    jitter=0.2, 
    ax=axes[1, 1]
)
axes[1, 1].set_title('4. Spending Score by Department (Boxplot)', fontsize=13, fontweight='bold')
axes[1, 1].set_xlabel('Department')
axes[1, 1].set_ylabel('Spending Score (1-100)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('Data_Visualization_Dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```
